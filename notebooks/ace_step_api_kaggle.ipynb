# ACE-Step API no Kaggle

Sobe a API FastAPI deste projeto usando o backend real (`MUSIC_BACKEND=acestep`)
dentro de um notebook Kaggle com GPU, e expoe publicamente via ngrok.

**Antes de rodar:** ative um acelerador de GPU em Notebook Settings (T4 ou P100),
e tenha em maos um authtoken gratuito do ngrok (https://dashboard.ngrok.com/get-started/your-authtoken).

````python
import os

os.environ["MUSIC_BACKEND"] = "acestep"
os.environ["ACE_STEP_CHECKPOINT_PATH"] = ""  # preencha se usar um checkpoint customizado
os.environ["PORT"] = "8000"
````

````python
import subprocess

server_process = subprocess.Popen(
    ["uv", "run", "uvicorn", "app.main:app", "--host", "0.0.0.0", "--port", "8000"],
)
print(f"uvicorn iniciado com PID {server_process.pid}")
````

````python
from pyngrok import ngrok, conf

conf.get_default().auth_token = "COLE_SEU_NGROK_AUTHTOKEN_AQUI"
public_tunnel = ngrok.connect(8000, "http")
print(f"API publica em: {public_tunnel.public_url}")
````

## Testando de fora do Kaggle

Use a URL impressa acima no lugar de `http://127.0.0.1:8000` nos exemplos do `README.md`, por exemplo:

````python
# Rode esta celula para encerrar o tunel e o servidor ao terminar a sessao.
ngrok.disconnect(public_tunnel.public_url)
server_process.terminate()
````
